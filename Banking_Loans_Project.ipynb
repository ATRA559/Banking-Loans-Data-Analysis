### Dataset Understanding

The Banking Loans Dataset contains information about loan accounts, customer characteristics, loan terms, credit risk indicators, and repayment-related metrics.

The dataset is used to assess data quality, perform data cleaning, explore patterns, and answer business-related questions using data analysis and visualization.

### Dataset:

| Column                    | Description                                                      |
| ------------------------- | ---------------------------------------------------------------- |
| `record_id`               | Unique identifier for each record in the dataset                 |
| `loan_id`                 | Identifier of the loan account                                   |
| `customer_age`            | Age of the customer in years                                     |
| `city`                    | City where the customer or loan account is located               |
| `branch`                  | Bank branch associated with the loan account                     |
| `employment_status`       | Employment status of the customer                                |
| `annual_income_egp`       | Customer's annual income in Egyptian Pounds (EGP)                |
| `loan_type`               | Type of loan associated with the account                         |
| `loan_purpose`            | Purpose for which the loan was taken                             |
| `loan_amount_egp`         | Original loan amount in Egyptian Pounds (EGP)                    |
| `interest_rate_pct`       | Interest rate applied to the loan, expressed as a percentage     |
| `term_months`             | Loan repayment term in months                                    |
| `credit_score`            | Customer's credit score used as an indicator of creditworthiness |
| `debt_to_income_pct`      | Customer's debt-to-income ratio, expressed as a percentage       |
| `risk_grade`              | Risk classification assigned to the loan or customer             |
| `origination_date`        | Date on which the loan was originated                            |
| `application_channel`     | Channel through which the loan application was submitted         |
| `loan_status`             | Current status of the loan account                               |
| `monthly_installment_egp` | Monthly loan installment amount in Egyptian Pounds (EGP)         |
| `remaining_balance_egp`   | Outstanding loan balance remaining in Egyptian Pounds (EGP)      |

##### **Duplicate validation note:**
During the cleaning process, 70 unique `loan_id` values were found to have duplicates, representing 140 rows. A comparison with the original dataset showed 8 missing cells within these duplicated records: 2 in `city`, 1 in `employment_status`, 3 in `annual_income_egp`, and 2 in `loan_purpose`. Six of these missing values were present in only one record of a duplicated `loan_id`, so the imputation temporarily created differences between otherwise duplicated records.
These 6 values were restored using the corresponding original values from their duplicate records. The remaining 2 missing `loan_purpose` values were missing in both duplicated records, so replacing both with `Unknown` preserved their consistency.
The duplicate records were then rechecked, confirming that the imputation-related differences had been resolved without requiring the cleaning process to be repeated.

##### **Duplicate records handling note:**
The duplicate `loan_id` investigation identified 70 duplicated loan IDs, each appearing exactly twice. After validating the records and excluding the `record_id` column, the duplicated records were found to contain identical loan information. Since `record_id` is only a record identifier and does not represent different loan information, one copy of each duplicated `loan_id` can be retained and the redundant duplicate record can be removed.


```python
# Remove redundant duplicate loan records
df = df.drop_duplicates(subset="loan_id", keep="first")

# Remove the record-id identifier
df = df.drop(columns="record_id")

# Reset the DataFrame index
df = df.reset_index(drop=True)
```
